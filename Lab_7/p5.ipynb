Suppose the campus mess collects ratings from 1000 students over a 7-day week (represented as a 1000 × 7 array of “random” integers between 1 and 5). Using aggregation functions along specific axes,

```python
import numpy as np
rng = np.random.default_rng(seed=303)
ratings = rng.integers(1,6,size=(1000,7))
ratings
```

Compute the average rating for each day of the week.

```python
avg_rating_perday = np.average(ratings,axis = 0)
avg_rating_perday
```

Compute the average rating for each student across the week.

```python
avg_rating_of_student = np.average(ratings,axis = 1)
avg_rating_of_student.shape
```

Find the index of the day (0 to 6) with the highest average rating

```python
max_rating_day = np.argmax(avg_rating_perday)
max_rating_day,avg_rating_perday[max_rating_day]
```

Standardize the ratings of each student so that every student’s
rating across the week has a mean of 0 and a standard deviation of 1.
How do you deal with the case where a student has the same rating

```python
mean_rating = np.mean(ratings,axis=0)
std_dev = np.std(ratings,axis=0)

normalised_ratings = (ratings - mean_rating)/std_dev
print(normalised_ratings.shape)
new_mean = np.mean(normalised_ratings,axis=0)
new_std_dev = np.std(normalised_ratings,axis=0)
print(new_mean)
print(new_std_dev)

# import numpy as np

# # 1. Calculate the initial mean and standard deviation
# mean_rating = np.mean(ratings, axis=0)
# std_dev = np.std(ratings, axis=0)

# # 2. Create the mask and filter out zero-variance columns
# variance_mask = std_dev > 0
# filtered_ratings = ratings[:, variance_mask]

# # 3. Normalize only the remaining useful columns
# normalised_ratings = (filtered_ratings - mean_rating[variance_mask]) / std_dev[variance_mask]
# print("New Shape:", normalised_ratings.shape)  # The column count will decrease if any std was 0

# # 4. Check the new mean and standard deviation of the processed columns
# new_mean = np.mean(normalised_ratings, axis=0)
# new_std_dev = np.std(normalised_ratings, axis=0)

# print("New Means (should be ~0):", new_mean)
# print("New Std Devs (should be 1):", new_std_dev)
```
